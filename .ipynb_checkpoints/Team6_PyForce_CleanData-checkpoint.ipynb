**Step 7-Preview the content of the Data file**

```python
print(Demo_clean.head())
print(Demo_clean.shape)
print(Demo_clean.columns)
print(Demo_clean.tail())
```

**Step 8-Inspection of data. Checking for datatypes,missing values,duplicate rows and counting rows**

```python
print(Demo_clean.dtypes, "\n")
print("Missing values:\n", Demo_clean.isna().sum(), "\n")
print("Duplicate rows:", Demo_clean.duplicated().sum())
for col in ["gender", "occupation", "agecat"]:
    print("\n", Demo_clean[col].value_counts(dropna=False))
Demo_clean[["weight", "height", "bmi"]].describe().round(2)
```

**Capitalise Coloumn Names**

```python
Demo_clean = Demo_clean.rename(columns={
   
    "agecat": "age_category",
})
```

**Step 9-Coloumn Text Formatting**

```python
for col in ["gender", "occupation", "age_category"]:
    Demo_clean[col] = Demo_clean[col].astype("string").str.strip()

Demo_clean["gender"] = Demo_clean["gender"].str.title()
```

**Step 10-Remove rows with no use data**

```python
before = len(Demo_clean)
Demo_clean = Demo_clean.dropna(subset=["gender", "weight", "height", "age_category"], how="all")
print(f"Dropped {before - len(Demo_clean)} empty row(s)")
```

**Step-11 Rename Coloumn. Renamed inpatient_number to patient_id to match with every other tables. Changed weight to weight_kg and height to height_meters for better coloumn read understanding.**

```python
Demo_clean = Demo_clean.rename(columns={
    "inpatient_number": "patient_id",
    "weight": "weight_kg",
    "height": "height_meters",
})
```

**Step 12-Check duplicate rows in patient_id**

```python
print("Duplicate rows:", Demo_clean.duplicated().sum())
print("Duplicate patient_ids:", Demo_clean["patient_id"].duplicated().sum())
Demo_clean = Demo_clean.drop_duplicates(subset="patient_id")
```

**Step 12-BMI to 2 decimal points for better reading**

```python
Demo_clean["bmi"] = Demo_clean["bmi"].round(2)
```

**Step 13-Occupation-Format and Empty Fields.Labels are standardised. Blanks are set to Inknown,not Others. Others is a real category in the data.**

```python
blank_occ = Demo_clean[Demo_clean["occupation"].isna()]
print("Blank occupations:", len(blank_occ))
print(blank_occ["gender"].value_counts(normalize=True).round(2).to_dict())
print(blank_occ["age_category"].value_counts(normalize=True).round(2).head(3).to_dict())

Demo_clean["occupation"] = Demo_clean["occupation"].replace({
    "UrbanResident": "Urban Resident", "farmer": "Farmer", "worker": "Worker"})
Demo_clean["occupation"] = Demo_clean["occupation"].fillna("Unknown")
Demo_clean["occupation"].value_counts()
```

**Step 14 -Age Category remove overlaps.The blank rows lean male(about two thirds,versus 42% overall) and sit mostly in the 60 to 89 age range,like the rest of the cohort. With only 27 rows that is not enough to infer an occupation,so no value is guessed.**

age_map = {
    "21-29": "21-29", "29-39": "30-39", "39-49": "40-49", "49-59": "50-59",
    "59-69": "60-69", "69-79": "70-79", "79-89": "80-89", "89-110": "90-110",
}
Demo_clean["age_category"] = Demo_clean["age_category"].map(age_map)
Demo_clean["age_category"] = pd.Categorical(
    Demo_clean["age_category"], categories=list(age_map.values()), ordered=True)
Demo_clean["age_category"].value_counts().sort_index()

```python
age_categories = [
    "21-29",
    "30-39",
    "40-49",
    "50-59",
    "60-69",
    "70-79",
    "80-89",
    "90-110"
]

age_map = {
    "21-29": "21-29",
    "29-39": "30-39",
    "39-49": "40-49",
    "49-59": "50-59",
    "59-69": "60-69",
    "69-79": "70-79",
    "79-89": "80-89",
    "89-110": "90-110"
}

# Only map the original categories
Demo_clean["age_category"] = (
    Demo_clean["age_category"]
    .replace(age_map)
)

Demo_clean["age_category"] = pd.Categorical(
    Demo_clean["age_category"],
    categories=age_categories,
    ordered=True
)

Demo_clean["age_category"].value_counts().sort_index()
```
