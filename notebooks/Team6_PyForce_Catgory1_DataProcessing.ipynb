```python
import pandas as pd
```

```python
df1 = pd.read_csv("../data/cleaned/cardiac_complications_cleaned.csv")
df2 = pd.read_csv("../data/cleaned/demography_cleaned.csv")
df3 = pd.read_csv("../data/cleaned/hospitalization_discharge_cleaned.csv")
df4 = pd.read_csv("../data/cleaned/labs_cleaned.csv")
df5 = pd.read_csv("../data/cleaned/patienthistory_cleaned.csv")
df6 = pd.read_csv("../data/cleaned/responsivenes_cleaned.csv")
df7 = pd.read_csv("../data/cleaned/patient_prescriptions_cleaned.csv")
```

Step 5 — Check the PatientID column

```python
for name, df in {
    "df1": df1,
    "df2": df2,
    "df3": df3,
    "df4": df4,
    "df5": df5,
    "df6": df6,
    "df7": df7
    
}.items():
    print(name, df.columns.tolist())
```

Step 6 — Check each file

```python
for name, df in {
    "df1": df1,
    "df2": df2,
    "df3": df3,
    "df4": df4,
    "df5": df5,
    "df6": df6,
    "df7": df7
}.items():
    
    print(
        name,
        "| Rows:", len(df),
        "| Unique Patients:", df["patient_id"].nunique(),
        "| Missing PatientID:", df["patient_id"].isna().sum(),
        "| Duplicate PatientID:", df["patient_id"].duplicated().sum()
    )
```

Step 7 — Choose your master file

```python
final_df = df1.copy()
```

Step 8 — Merge the second CSV

```python
final_df = final_df.merge(
    df2,
    on="patient_id",
    how="left",
    validate="one_to_one"
)
```

Step 9 — Merge the third CSV

```python
final_df = final_df.merge(
    df3,
    on="patient_id",
    how="left",
    validate="one_to_one"
)
```

Step 10 — Merge the fourth CSV

```python
final_df = final_df.merge(
    df4,
    on="patient_id",
    how="left",
    validate="one_to_one"
)
```

Step 11 — Merge the fifth CSV

```python
final_df = final_df.merge(
    df5,
    on="patient_id",
    how="left",
    validate="one_to_one"
)
```

Step 12 — Merge the sixth CSV

```python
final_df = final_df.merge(
    df6,
    on="patient_id",
    how="left",
    validate="one_to_one"
)
```

Step 12 — Merge the seventh CSV

```python
final_df = final_df.merge(
    df7,
    on="patient_id",
    how="left",
    validate="one_to_one"
)
```

Step 13 — Check your final result

```python
print("Final rows:", len(final_df))
print("Unique patients:", final_df["patient_id"].nunique())
print("Duplicate PatientIDs:", final_df["patient_id"].duplicated().sum())
print("Missing PatientIDs:", final_df["patient_id"].isna().sum())
```

Step 15 — Check columns

```python
print(final_df.shape)
print(final_df.columns.tolist())
[col for col in final_df.columns if col.endswith("_x") or col.endswith("_y")]
```

Step 16  — Check for duplicate column names

```python
print(final_df.columns[final_df.columns.duplicated()].tolist())
```

Step 17 — Save your final CSV

```python
final_df.to_csv(
    "../data/final/Team6_PyForce_cleaned_data.csv",
    index=False
)
```
